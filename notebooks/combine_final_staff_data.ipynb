# Combine municipality and gaupalika staff data

Standardize equivalent column names, concatenate the datasets row-wise, and order the columns with geographic information first. Pandas keeps the union of their columns; fields that occur in only one dataset will be blank for rows from the other. This prepares common name, province, and district columns for a later join but does not perform a key-based merge.

```python
from pathlib import Path

import pandas as pd

data_dir = Path.cwd().parent / "data"  
gaupalika_file = data_dir / "Gaupalika_staff_final.xlsx"
municipality_file = data_dir / "Municipality_staff_final.xlsx"
combined_file = data_dir / 'Officials_combined' / "Combined_staff_final.xlsx"
```

```python
gaupalika_df = pd.read_excel(gaupalika_file)
municipality_df = pd.read_excel(municipality_file)

print("Gaupalika shape:", gaupalika_df.shape)
print("Gaupalika columns:", gaupalika_df.columns.tolist())
print("Municipality shape:", municipality_df.shape)
print("Municipality columns:", municipality_df.columns.tolist())
```

```python
gaupalika_df = gaupalika_df.rename(
    columns={"Local Level Name": "Municipality/Gaupalika", "Desgnations": "Designations"}
)
municipality_df = municipality_df.rename(
    columns={"Unnamed: 0": "Municipality/Gaupalika", "province": "Province", "district": "District"}
)

print("Gaupalika columns after rename:", gaupalika_df.columns.tolist())
print("Municipality columns after rename:", municipality_df.columns.tolist())
```

```python
print("Gaupalika columns:", gaupalika_df.columns.tolist())
print("Municipality columns:", municipality_df.columns.tolist())
```

```python
combined_df = pd.concat(
    [gaupalika_df, municipality_df],
    ignore_index=True,
    sort=False,
)

region_and_name_columns = [
    "Province",
    "District",
    "Municipality/Gaupalika",
    "Names",
]
remaining_columns = [
    column for column in combined_df.columns
    if column not in region_and_name_columns
]
combined_df = combined_df[region_and_name_columns + remaining_columns]

print("Combined shape:", combined_df.shape)
print("Combined columns:", combined_df.columns.tolist())
display(combined_df.head())
```

```python
combined_df.to_excel(combined_file , index=False)
print(f"Saved combined data to: {combined_file}")
```
