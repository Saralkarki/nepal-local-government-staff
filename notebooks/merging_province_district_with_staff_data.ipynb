```python
import pandas as pd
from pathlib import Path

```

```python
data_dir = Path.cwd().parent / "data"
```

```python
data_dir
```

```python
    # read the gaupalika and the municipality files
df_gaupalika = pd.read_excel(data_dir / 'Gaupalika_Municipality_url_list' / 'updated_gaupalika_contacts.xlsx')
df_muni = pd.read_excel(data_dir / 'Gaupalika_Municipality_url_list' / 'updated_Muan_contacts.xlsx')
```

```python
# Read the staff data for gauaplika and municipality
df_staff_gau = pd.read_excel(data_dir /'Officials'/'final_combined_data_gaupalika_cleaned.xlsx')
df_staff_mun = pd.read_excel(data_dir /'Officials'/'final_combined_data_municipality_cleaned.xlsx')
```

```python
# look at the shape before merge
print(df_gaupalika.shape, df_staff_gau.shape)
```

```python
print(df_muni.shape, df_staff_mun.shape)
```

```python
df_muni.columns, df_staff_gau.columns
```

```python
df_muni.rename(columns={'working_url': 'source_url'}, inplace=True)
df_gaupalika.rename(columns={'working_url': 'source_url'}, inplace=True)
```

```python
# only selected columns for gaupalika and municipality 
df1 = df_gaupalika[['Province', 'Local Level Name', 'District', 'Email',
        'Contact Phone', 'municipality_type',
        'source_url',]]
df2 = df_muni[['municipality_name', 'province', 'district', 'website',
        'source_url']]
```

```python
# merge municipality
muni_merged = df2.merge(df_staff_mun, on = 'source_url')
```

```python
# the gaupalika df had duplicated values
df1[df1.duplicated('source_url', keep=False)].sort_values('source_url')
```

```python
df1 = df1.drop_duplicates(subset='source_url', keep='first')
```

```python
df1.shape
```

```python
gaupalika_merged = df_staff_gau.merge(df1, on='source_url', how='left')
```

```python
gaupalika_merged.shape
```

```python
## export the files
muni_merged.to_excel(data_dir /'Officials'/ "Municipality_staff_final.xlsx", index = False)
gaupalika_merged.to_excel(data_dir /'Officials'/ "Gaupalika_staff_final.xlsx", index = False)

```
